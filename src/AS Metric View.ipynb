```python
%sql
USE CATALOG IDENTIFIER(:catalog);
USE IDENTIFIER(:schema);

CREATE OR REPLACE VIEW opportunities_metric_view__DDL
WITH METRICS LANGUAGE YAML AS $$
version: '1.1'
source: databricks_simulated_australia_sales_and_opportunities_data.v01.opportunities
joins:
  - name: customers
    source: databricks_simulated_australia_sales_and_opportunities_data.v01.customers
    on: source.customerid = customers.customerid
dimensions:
  - name: city
    expr: customers.city
  - name: state
    expr: customers.state
measures:
  - name: Open Opportunities
    expr: SUM(amount) FILTER (WHERE left(source.phase,1) in ('0','1'))
    comment: Open Opportunities
    display_name: Open Opportunities
  - name: Won Opportunities
    expr: SUM(amount) FILTER (WHERE left(source.phase,1) ='2')
    comment: Won Opportunities
    display_name: Won Opportunities
$$;
```
